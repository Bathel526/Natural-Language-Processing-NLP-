## *n*-gramy
- n-gram to ciąg n elementów występujących w określonym porządku
- elementami przeważnie są słowa, choć mogą to być także sylaby, litery, itp.
- przykłady 2-gramów: (on, idzie), (idzie, do), (nie, pierwszy)
- przykłady 3-gramów: (on, idzie, tam), (Francis, Ford, Coppola)
- n-gramy pozwalają zachować informację o 
kolejności wyrazów – np. (nie, pierwszy) zachowa 
więcej informacji niż tokeny „nie” i „pierwszy”
- uwzględnianie n-gramów może znacząco 
zwiększyć słownik (zbiór tokenów)
- w praktyce odfiltrowuje się n-gramy występujące w 
zbyt małej lub zbyt dużej liczbie dokumentów

```python
s = 'The United Kingdom is a country in Northwestern Europe.'
```

```python
from nltk.util import ngrams

tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(s)
list(ngrams(tokens, 2))
```

## Stop listy

stop listy zawierają słowa, które występują 
często w danym języku, np. w jęz. angielskim:
- spójniki (and, or)
- rodzajniki (a, an, the)
- zaimki (this),

cd

- podczas tokenizacji słowa ze stop listy są pomijane
- argumentem za pomijaniem słów ze stop listy jest to, że nie wnoszą istotnej informacji na temat frazy
- jednak występując w n-gramie mogą dostarczać istotną informację na temat relacji między słowami
- przykład utraty części informacji z dwóch 4-gramów przy stop liście zawierającej słowa on oraz off:
		- (turn, on, the, lights) → (turn, lights)
        - (turn, off, the, lights) → (turn, lights)

## Reprezentacja 1 z n
- Reprezentacja 1 z n przechowuje tekst w postaci ciągu wektorów zero-jedynkowych
- Każdy wektor reprezentuje jeden token
- Wektory są indeksowane tokenami (pozycją tokena w słowniku tokenów)
- Na pozycji odpowiadającej tokenowi występuje wartość 1, pozostałe pozycje to 0

cd
- nie traci żadnej informacji z tekstu (jeżeli żadnej informacji nie utracił tokenizator)
- ma wysoką złożoność pamięciową– jest stosowana m.in. w sieciach neuronowych
Dokument i jego tokenizacja:

```python
s = 'Be yourself; everyone else is already taken.' # Oscar Wilde
tokenizer = TreebankWordTokenizer()
tokens = tokenizer.tokenize(s) 
tokens
```

Utworzenie słownika (z eliminacją duplikatów):

```python
vocabulary = sorted(list(set(tokens)))
vocabulary
```

Utworzenie reprezentacji 1 z *n* w tablicy NumPy:

```python
# !pip install numpy
```

```python
import numpy as np

one_hot_sentence = np.zeros([len(tokens), len(vocabulary)])
for i, word in enumerate(tokens):
    one_hot_sentence[i, vocabulary.index(word)] = 1
one_hot_sentence
```

Czytelniejsza wersja w postaci ramki danych Pandas:

```python
# !pip install pandas
```

```python
import pandas as pd
df = pd.DataFrame(one_hot_sentence.astype(int), columns=vocabulary)
df
```

## Reprezentacja BoW
- Reprezentacja BoW (Bag of Words – worek słów)
- przechowuje tekst w postaci wektora
- Każdy wektor reprezentuje jeden dokument
- Wektory są indeksowane tokenami
- Na pozycji odpowiadającej tokenowi występuje częstość występowania tego tokena w dokumencie

cd
- ma mniejszą złożoność pamięciową niż 1 z n
- nie zachowuje informacji o kolejności słów
- wykorzystywana w klasyfikacji dokumentów, gdzie każdy token reprezentuje cechę danych wejściowych klasyfikatora

```python
# !pip install scikit-learn
```

```python
from sklearn.feature_extraction.text import CountVectorizer

s = 'Well done is better than well said' # Benjamin Franklin
cv = CountVectorizer()
bow = cv.fit_transform([s])
cv.get_feature_names_out()
```

```python
bow.toarray()
```

```python
df = pd.DataFrame(bow.toarray(), columns=cv.get_feature_names_out())
df
```

Tworzenie reprezentacji dla wielu dokumentów

```python
corpus = [
    'Well done is better than well said.', # Benjamin Franklin
    'Be yourself; everyone else is already taken.', # Oscar Wilde
    'Only a life lived for others is a life worthwhile.' # Albert Einstein
]

cv = CountVectorizer()
bow = cv.fit_transform(corpus)
cv.get_feature_names_out()
```

```python
bow.toarray()
```

```python
df = pd.DataFrame(bow.toarray(), columns=cv.get_feature_names_out())
pd.set_option('display.max_columns', 10)
df
```

Bardziej "oszczędna" wersja słownikowa:

```python
# dla pierwszego wektora BoW w tabeli
i = 0
# indeksy tylko dodatnich wartości częstości
index = df.iloc[0,]>=1
# słownik dla pierwszego wektora BoW
dict(zip(df.columns[index], df.loc[i,index]))
```

# Redukcja słownika

- Słownik (zbiór wszystkich tokenów) 
oznaczamy symbolem V 
- Rozmiar słownika  oznaczamy |V| 
- W obu omawianych reprezentacjach rozmiar 
słownika ma wpływ na rozmiar struktury danych

Aby zmniejszyć ten rozmiar stosuje się metody redukcji słownika minimalizujące utratę informacji:
- ujednolicanie wielkości liter
- Stemming
- lematyzację 

## Ujednolicanie wielkości liter
zmiana na małe litery tylko pierwszego wyrazu w zdaniu

zaleta:
- zachowanie nazw własnych

wada:
- utrata informacji o nazwie własnej na pierwszej pozycji w zdaniu
- „Biedronka i Żabka konkurują w całej Polsce.”

```python
s = 'The United Kingdom is a country in Northwestern Europe.'
s.lower()
```

## Stemming
- Stemming polega na usunięciu z wyrazu 
końcówki przy pozostawieniu tematu lub 
rdzenia
- Przykład: wyrazy program, programming, 
programmer, programs po zastosowaniu 
stemmingu sprowadzają się do słowa program


```python
from nltk.tokenize import TreebankWordTokenizer
from nltk.stem.porter import PorterStemmer

s = 'The United Kingdom is a country in Europe'
tokenizer = TreebankWordTokenizer()
stemmer = PorterStemmer()
for t in tokenizer.tokenize(s):
    print(stemmer.stem(t), end=', ')
```

# Lematyzacja
- Lematyzacja polega na grupowaniu słów odnoszących mających to samo lub podobne znaczenie (lemat)
- Lematyzator wykorzystuje wiedzę o synonimach słowa, jak np. WordNet


```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

s = 'was are be were am'
tokenizer = TreebankWordTokenizer()
for t in tokenizer.tokenize(s):
    print(lemmatizer.lemmatize(t), end=', ')
```

```python
for t in tokenizer.tokenize(s):
    print(lemmatizer.lemmatize(t, pos='v'), end=', ') # verb
```
